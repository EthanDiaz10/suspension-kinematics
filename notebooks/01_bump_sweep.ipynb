# Bump sweep walkthrough
Load the baseline hardpoints, solve ±50 mm of wheel travel, and look at the curves.
Try editing a hardpoint below and re-running to see how the curves move.

```python
import suspkin as sk

hp = sk.load_hardpoints('../data/baseline_fsae_front.json')
hp.chassis['UAF'], hp.chassis['UAA']
```

```python
# Experiment: raise the upper-arm chassis pivots by 10 mm and compare.
# hp.chassis['UAF'][2] += 10
# hp.chassis['UAA'][2] += 10

poses = sk.bump_sweep(hp, travel_mm=50, step_mm=1)
metrics = sk.compute_metrics(poses)
sk.summary(metrics)
```

```python
%matplotlib inline
sk.plot_sweep(metrics, title=hp.name);
```

```python
sk.plot_front_view(poses);
```
